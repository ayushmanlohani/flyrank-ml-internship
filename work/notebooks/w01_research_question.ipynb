## 1. My lane is content freshness.

*I want to investigate whether older or less recently updated pages are more likely to be declining in search performance.*

```python
import pandas as pd

# Load dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Create the boolean label for declining pages
df['is_declining_label'] = df['trend_direction'] == 'down'

# Calculate and display the mean metrics
metrics = df.groupby('is_declining_label')[['content_age_days', 'days_since_last_update']].mean()
print("Mean values for content age and freshness:")
print(metrics)
```

## 2. The question: decision, action, cost of a wrong call

- **What decision does this improve?** It improves the decision of **which older or declining pages should be refreshed/updated first** by a content editor.
- **Who acts on it?** The **Content Editorial Team / Editors** who will prioritize and update the flagged pages.
- **What does a wrong recommendation cost?**
  - *False Positive:* Wasted editor hours spent rewriting pages that didn't need updates.
  - *False Negative:* Missing a declining page, leading to lost traffic, impressions, and revenue.

```python
print("Declining pages:", df["is_declining_label"].sum())
print("Total pages:", len(df))
print("Declining rate:", df["is_declining_label"].mean())
```

## 3. Quick look at the data (2-3 real numbers)

From our initial run on the anonymized data, we found:
1. **Freshness Impact:** Pages that are declining in performance (`trend_direction == 'down'`) have not been updated for **49.2 days** on average, compared to **42.4 days** for stable or growing pages.
2. **Overall Age:** Declining pages have a mean overall content age of **236.2 days**, whereas non-declining pages have a mean age of **279.8 days**.

```python
print(df.groupby("is_declining_label")[["content_age_days", "days_since_last_update"]].mean())
```

## 4. Careful words: what I can and can't claim

- **What I can claim:** We can claim **directional and decision-support relationships** (e.g., *"Pages that haven't been updated for over 49 days are directionally more likely to show performance decline than those updated within 42 days"*).
- **What I can't claim:** We cannot claim **causal proof** (e.g., we cannot prove that the lack of updates is the sole cause of decline) or claim that we are **'predicting Google'** or its search ranking algorithms directly.

```python
print("Mean days since update:")
print(df.groupby("is_declining_label")["days_since_last_update"].mean())

print("\nMean content age:")
print(df.groupby("is_declining_label")["content_age_days"].mean())
```
